# Задача 1: Класифікація текстів

## 🧾 Як представити текст: від слів до чисел

Машини не працюють безпосередньо з текстом — їм потрібні **числові вектори**.  
Тому перший крок у NLP — **перетворення тексту в числове представлення**. Існує кілька підходів:

---

### ✂️ 1. Токенізація

**Токенізація** — це розбиття тексту на окремі елементи: слова, речення або підслова.

**Приклад:**
“Я люблю машинне навчання” → [“Я”, “люблю”, “машинне”, “навчання”]

---

### 🧮 2. Bag of Words (BoW)

Модель BoW представляє текст як **частоти слів**, ігноруючи порядок.

**Приклад:**

| Слово       | Частота |
|-------------|---------|
| машинне     | 1       |
| навчання    | 1       |
| люблю       | 1       |
| Я           | 1       |

Недоліки:

- Не враховує порядок слів
- Велика розмірність для великих словників

---

### 🔍 3. TF-IDF (Term Frequency — Inverse Document Frequency)

TF-IDF зменшує вагу "поширених" слів (як "і", "це", "я") і підсилює важливіші.

**Формула:**

$$TF-IDF = TF * log(N / DF)$$

де:
- **TF** — частота терміну в документі  
- **DF** — кількість документів з цим терміном  
- **N** — загальна кількість документів

TF-IDF — хороший компроміс між простотою BoW і складністю сучасних embeddings.
 

```python
from sklearn.feature_extraction.text import CountVectorizer, TfidfVectorizer  # Імпортуємо CountVectorizer для BoW і TfidfVectorizer для TF-IDF

# Приклад корпусу
texts = [  # Створюємо список текстових документів
    "Я люблю машинне навчання",  # Перший документ
    "Машинне навчання — це цікаво",  # Другий документ
    "Я вивчаю машинне навчання щодня"  # Третій документ
]

# --- Bag of Words ---
bow_vectorizer = CountVectorizer()  # Створюємо об'єкт для перетворення тексту в BoW
bow_matrix = bow_vectorizer.fit_transform(texts)  # Навчаємо векторизатор на текстах і перетворюємо їх у матрицю

print("🎒 Bag of Words (BoW):")  # Виводимо заголовок результату BoW
print(bow_vectorizer.get_feature_names_out())  # Виводимо всі слова, які стали ознаками
print(bow_matrix.toarray())  # Перетворюємо розріджену матрицю у звичайний масив і виводимо її

# --- TF-IDF ---
tfidf_vectorizer = TfidfVectorizer()  # Створюємо об'єкт для TF-IDF перетворення
tfidf_matrix = tfidf_vectorizer.fit_transform(texts)  # Навчаємо векторизатор і перетворюємо тексти у TF-IDF матрицю

print("\n📊 TF-IDF:")  # Виводимо заголовок результату TF-IDF
print(tfidf_vectorizer.get_feature_names_out())  # Виводимо слова, для яких обчислено TF-IDF
print(tfidf_matrix.toarray())  # Перетворюємо розріджену матрицю у звичайний масив і виводимо її
```

```python
# ----------------- ЦЕЙ КОД НЕ ПРАЦЮЄ У МЕНЕ -----------------
# Змінив версію Python і почало працювати: Python 3.12

from sklearn.datasets import fetch_20newsgroups  # Імпортуємо функцію для завантаження датасету 20 Newsgroups
from sklearn.feature_extraction.text import CountVectorizer, TfidfVectorizer  # Імпортуємо методи BoW та TF-IDF
from sklearn.linear_model import LogisticRegression  # Імпортуємо логістичну регресію для класифікації тексту
from sklearn.model_selection import train_test_split  # Імпортуємо функцію для поділу даних на train і test
from sklearn.metrics import classification_report  # Імпортуємо звіт з precision, recall та F1-score

# 1. Завантажуємо текстовий датасет (2 категорії для простоти)
categories = ['sci.space', 'rec.sport.hockey']  # Вибираємо лише дві категорії: космос і хокей
data = fetch_20newsgroups(  # Завантажуємо датасет 20 Newsgroups
    subset='all',  # Використовуємо всі доступні дані: train + test
    categories=categories,  # Завантажуємо тільки вибрані дві категорії
    remove=('headers', 'footers', 'quotes')  # Видаляємо заголовки, підписи та цитати з текстів
)

# print("Hello my friends")  # Тестовий print, зараз закоментований і не виконується
data.data[0]  # Виводимо перший текст із датасету
```

```python
from sklearn.linear_model import LogisticRegression  # Імпортуємо модель логістичної регресії для класифікації
from sklearn.model_selection import train_test_split  # Імпортуємо функцію для поділу даних на train і test
from sklearn.feature_extraction.text import CountVectorizer, TfidfVectorizer  # Імпортуємо BoW та TF-IDF векторизатори
from sklearn.metrics import classification_report  # Імпортуємо звіт для оцінки якості моделі

# 2. Ділимо на train/test
X_train, X_test, y_train, y_test = train_test_split(  # Ділимо тексти та їх класи на навчальну і тестову частини
    data.data,  # Передаємо всі текстові документи
    data.target,  # Передаємо правильні класи для кожного документа
    test_size=0.3,  # 30% даних використовуємо для тестування, 70% для навчання
    random_state=42  # Фіксуємо випадковість, щоб при повторному запуску отримати той самий поділ
)

# 3. Векторизація тексту (спробуй BoW або TF-IDF)
# vectorizer = CountVectorizer()  # Створюємо BoW-векторизатор, який рахує кількість появи слів
vectorizer = TfidfVectorizer()  # Створюємо TF-IDF-векторизатор, який оцінює важливість слів

X_train_vec = vectorizer.fit_transform(X_train)  # Навчаємо TF-IDF на train-текстах і перетворюємо їх у числові вектори
X_test_vec = vectorizer.transform(X_test)  # Перетворюємо test-тексти за словником, отриманим тільки з train

# 4. Тренування моделі
clf = LogisticRegression(max_iter=1000)  # Створюємо модель логістичної регресії з максимумом 1000 ітерацій
clf.fit(X_train_vec, y_train)  # Навчаємо модель на векторизованих текстах та правильних класах

# 5. Оцінка
y_pred = clf.predict(X_test_vec)  # Модель прогнозує класи для тестових текстів
print(classification_report(y_test, y_pred, target_names=data.target_names))  # Виводимо precision, recall, F1-score та accuracy
```

## ✂️ Стемінг і лемматизація

Перш ніж векторизувати текст, часто варто його **нормалізувати** — звести слова до базової форми.  
Це допомагає зменшити розмір словника та об'єднати варіанти одного слова.

---

### 🌱 Стемінг (Stemming)

**Стемінг** — це грубе обрізання слів до кореня (стему).  
Метод швидкий, але іноді "ламає" слова.

**Приклад:**
“навчаю”, “навчання”, “навчався” → “навч”

```python
from nltk.stem import PorterStemmer

stemmer = PorterStemmer()
print(stemmer.stem("learning"))     # → 'learn'
print(stemmer.stem("learned"))      # → 'learn'
```

## 🌳 Лемматизація (Lemmatization)

Лемматизація — це зведення слова до його словникової форми (леми).
Метод точніший за стемінг, але повільніший і потребує знання частини мови.

Приклад:
"навчаю", "навчання", "навчався" → "навчати"

```python
from nltk.stem import WordNetLemmatizer  # Імпортуємо лематизатор, який приводить слова до початкової форми
import nltk  # Імпортуємо бібліотеку NLTK для роботи з природною мовою

nltk.download('wordnet')  # Завантажуємо словник WordNet, необхідний для лематизації

lemmatizer = WordNetLemmatizer()  # Створюємо об'єкт лематизатора

print(lemmatizer.lemmatize("learning", pos="v"))  # Лематизуємо "learning" як дієслово → "learn"
print(lemmatizer.lemmatize("better", pos="a"))  # Лематизуємо "better" як прикметник → "good"
```

```python
import nltk  # Імпортуємо бібліотеку NLTK для роботи з текстом
from nltk.stem import PorterStemmer  # Імпортуємо Porter Stemmer для стемінгу слів
from nltk.tokenize import word_tokenize  # Імпортуємо токенізатор для розбиття речення на слова

nltk.download('punkt_tab')  # Завантажуємо дані, необхідні для роботи word_tokenize

stemmer = PorterStemmer()  # Створюємо об'єкт стемера Porter

# Функція для стемінгу речення
def apply_stemming(sentence):  # Створюємо функцію, яка отримує одне речення
    words = word_tokenize(sentence)  # Розбиваємо речення на окремі слова
    stemmed_words = [stemmer.stem(word) for word in words]  # Для кожного словамо його основу
    return ' '.join(stemmed_words)  # Об'єднуємо оброблені слова отримує назад у речення

# Припустимо, data.data — це список текстових документів
stemmed_train_data = [apply_stemming(sentence) for sentence in data.data]  # Застосовуємо стемінг до кожного документа

from sklearn.feature_extraction.text import TfidfVectorizer  # Імпортуємо TF-IDF векторизатор

vectorizer = TfidfVectorizer()  # Створюємо TF-IDF векторизатор
converted = vectorizer.fit_transform(stemmed_train_data)  # Навчаємо TF-IDF на оброблених текстах і перетворюємо їх у числову матрицю
print(converted.shape)  # Виводимо розмір отриманої TF-IDF матриці
```

# 🧩 Задача 2: Класифікація токенів

**Класифікація токенів** — це тип задачі в NLP, де модель присвоює **кожному токену в тексті** певну категорію або мітку.

---

### 📌 Типові задачі класифікації токенів:

| Задача                      | Приклад міток           | Для чого це |
|----------------------------|--------------------------|-------------|
| 🔤 **POS-tagging**         | `NOUN`, `VERB`, `ADJ`   | Частина мови |
| 🏷️ **NER**                | `B-PER`, `I-ORG`, `O`   | Іменовані сутності |
| 🧱 **Chunking**            | `B-NP`, `I-NP`, `O`      | Іменні групи |
| 🪓 **Segmentation**        | `B`, `I`, `E`, `S`       | Розбиття тексту |
| 🔐 **Masked Language Modeling** | `MASK` або індекс мітки | Передбачення токенів |

---

### 🤖 Як це працює?

1. **Текст розбивається на токени**
2. **Кожному токену призначається мітка**
3. **Модель навчається передбачати мітки для нових токенів**

---

### 📐 Формат даних

Зазвичай у форматі **послідовності** токенів і відповідних **міток**:

```python
import nltk  # Імпортуємо бібліотеку NLTK для роботи з природною мовою
from nltk.corpus import conll2002  # Імпортуємо датасет CoNLL-2002

# cat dog mouse
# [1 0 0]
# [0 1 0]

nltk.download('conll2002')  # Завантажуємо датасет CoNLL-2002, якщо його ще немає

# Load the train and test sets
train_sents = conll2002.tagged_sents('ned.train')  # Завантажуємо навчальні речення з нідерландського набору
test_sents = conll2002.tagged_sents('ned.testa')  # Завантажуємо тестові речення з нідерландського набору

# print("train_sents", len(train_sents))
# print("train_sents[0]", train_sents[0])

# print("test_sents", len(test_sents))
# print("test_sents[0]", test_sents[0])

# Extract word-tag pairs
def extract_word_tag_pairs(sentences):  # Створюємо функцію для отримання пар слово-тег
    processed_sentences = []  # Створюємо порожній список для оброблених речень

    for sentence in sentences:  # Перебираємо всі речення
        processed_sentences.append([(word, tag) for word, tag in sentence])  # Додаємо кожне слово разом із його тегом

    return processed_sentences  # Повертаємо список оброблених речень

train_data = extract_word_tag_pairs(train_sents)  # Обробляємо навчальний набір
test_data = extract_word_tag_pairs(test_sents)  # Обробляємо тестовий набір

print(len(train_data), len(test_data))  # Виводимо кількість речень у train та test

# Print the first few sentences to inspect the data format
print("First 5 training sentences:")  # Виводимо заголовок для навчальних речень

for i in range(min(5, len(train_data))):  # Перебираємо перші 5 речень або менше, якщо їх недостатньо
    print(train_data[i])  # Виводимо поточне навчальне речення

print("\nFirst 5 testing sentences:")  # Виводимо заголовок для тестових речень

for i in range(min(5, len(test_data))):  # Перебираємо перші 5 тестових речень або менше
    print(test_data[i])  # Виводимо поточне тестове речення
```

```python
# Імпортуємо модель логістичної регресії.
# Вона буде використовуватися для класифікації слів за їхніми тегами.
from sklearn.linear_model import LogisticRegression

# Імпортуємо функцію для розділення даних на навчальну та тестову вибірки.
from sklearn.model_selection import train_test_split

# Імпортуємо функцію для побудови звіту про якість класифікації.
# Звіт міститиме precision, recall, f1-score та support.
from sklearn.metrics import classification_report

# Імпортуємо DictVectorizer.
# Він перетворює словники ознак у числову матрицю,
# яку може використовувати машинна модель.
from sklearn.feature_extraction import DictVectorizer


# Об'єднуємо навчальні та тестові речення в один список.
# train_data містить навчальні речення.
# test_data містить тестові речення.
# Після цього all_data містить усі речення разом.
all_data = train_data + test_data


# Створюємо порожній список для початкових ознак.
# У нього будемо додавати словники з ознаками кожного слова.
X_raw = []


# Створюємо порожній список для правильних відповідей (міток).
# Тут будуть зберігатися POS-теги слів, наприклад:
# NN, VB, JJ, PRP тощо.
y = []


# Проходимо по кожному реченню у всьому наборі даних.
for sentence in all_data:

    # Проходимо по кожному слову та його тегу в поточному реченні.
    # word — саме слово.
    # tag — його правильний граматичний тег.
    for word, tag in sentence:

        # Створюємо словник ознак для поточного слова.
        # Наприклад:
        # {"word": "cat"}
        #
        # Поки що використовуємо тільки одну ознаку — саме слово.
        X_raw.append({"word": word})

        # Додаємо правильний тег цього слова до списку y.
        # Наприклад:
        # y = ["NN", "VB", "DT", ...]
        y.append(tag)


# Створюємо об'єкт DictVectorizer.
# sparse=True означає, що результат буде зберігатися
# у вигляді розрідженої матриці (sparse matrix).
vec = DictVectorizer(sparse=True)


# Перетворюємо словники X_raw у числову матрицю.
#
# Наприклад, якщо слова:
# {"word": "cat"}
# {"word": "dog"}
# {"word": "runs"}
#
# DictVectorizer створить приблизно такі ознаки:
#
# word=cat   word=dog   word=runs
#     1          0          0
#     0          1          0
#     0          0          1
#
# fit_transform() одночасно:
# 1. знаходить усі можливі слова;
# 2. створює для них числові ознаки;
# 3. перетворює X_raw у матрицю X_vec.
X_vec = vec.fit_transform(X_raw)


# Розділяємо весь набір даних на навчальну та тестову частини.
#
# X_vec — ознаки всіх слів.
# y — правильні POS-теги всіх слів.
#
# test_size=0.2 означає:
# 80% даних → навчання моделі
# 20% даних → перевірка моделі
#
# random_state=42 фіксує випадковість,
# щоб при повторному запуску отримувати такий самий поділ.
X_train, X_test, y_train, y_test = train_test_split(
    X_vec,
    y,
    test_size=0.2,
    random_state=42
)


# Створюємо модель логістичної регресії.
#
# LogisticRegression використовується для класифікації.
#
# max_iter=1000 дозволяє алгоритму виконати максимум
# 1000 ітерацій під час навчання.
#
# Це збільшує шанс, що модель встигне знайти оптимальні параметри.
clf = LogisticRegression(max_iter=1000)


# Навчаємо модель на навчальній вибірці.
#
# X_train — ознаки слів.
# y_train — правильні POS-теги.
#
# Модель намагається навчитися залежності:
#
# слово → POS-тег
#
# Наприклад:
# "dog" → NN
# "run" → VB
# "the" → DT
clf.fit(X_train, y_train)


# Використовуємо навчену модель для прогнозування тегів
# для тестової вибірки.
#
# X_test містить слова, які модель не використовувала
# під час навчання.
#
# y_pred — передбачені моделлю POS-теги.
y_pred = clf.predict(X_test)


# Виводимо детальний звіт про якість роботи моделі.
#
# y_test — правильні теги.
# y_pred — теги, передбачені моделлю.
#
# classification_report показує:
#
# precision — наскільки точними були передбачення класу;
# recall — яку частину правильних об'єктів модель знайшла;
# f1-score — середнє між precision та recall;
# support — кількість прикладів кожного класу.
print(classification_report(y_test, y_pred))
```

## 🧠 Word Embeddings — векторне представлення слів

**Word embeddings** — це спосіб представити слова як **щільні вектори** з десятками або сотнями вимірів, які зберігають **семантичну подібність** між словами.

---

### 🔍 Навіщо потрібні ембедінги?

Методи на кшталт `DictVectorizer` або one-hot перетворюють слова в розріджені та **повністю незалежні вектори**. Наприклад:
- "корова" → `[0, 0, 1, 0, 0]`
- "собака" → `[0, 1, 0, 0, 0]`

📉 Але модель не знає, що ці слова **схожі за змістом**.

🔼 З ембедінгами:
- "cow" → `[0.45, -1.2, ..., 0.87]`
- "dog" → `[0.42, -1.1, ..., 0.85]`

📐 Вектори схожі, бо значення близьке у семантичному просторі.

---

### 🧭 Властивості ембедінгів

- Схожі слова → близькі вектори
- Операції типу:  
  `king - man + woman ≈ queen`
- Побудовані на **контексті**, в якому зустрічається слово

---

### 🧰 Популярні типи ембедінгів

| Метод        | Характеристика                   |
|--------------|----------------------------------|
| **Word2Vec** | Навчає слова за контекстом       |
| **GloVe**    | Використовує статистику корпусу  |
| **FastText** | Працює з n-грамами, знає нові слова |
| **BERT**     | Контекстні ембедінги (динамічні) |

---

### 🧠 Архітектури Word2Vec

- **CBOW (Continuous Bag of Words)**:  
  Передбачає слово за контекстом  
  (напр., "I _ to school" → "go")

- **Skip-gram**:  
  Передбачає контекст за словом  
  (напр., "_ → ['I', 'to', 'school']")

```python
import numpy as np  # Підключаємо бібліотеку NumPy для роботи з масивами та числовими векторами
from sklearn.linear_model import LogisticRegression  # Імпортуємо модель логістичної регресії для класифікації
from sklearn.model_selection import train_test_split  # Імпортуємо функцію для розділення даних на train і test
from sklearn.metrics import classification_report  # Імпортуємо функцію для оцінювання якості класифікації

# Assuming 'train_data', 'test_data', 'glove_vectors' are defined from previous steps  # Припускаємо, що train_data, test_data та glove_vectors вже створені раніше

# Function to convert sentences to GloVe vectors and extract corresponding tags  # Функція перетворює речення на GloVe-вектори та окремо зберігає теги
def sentences_to_glove_features_and_tags(sentences, glove_model):  # Оголошуємо функцію, яка приймає речення та GloVe-модель
    features = []  # Створюємо список для зберігання векторів слів
    tags = []  # Створюємо список для зберігання тегів кожного слова
    vector_size = glove_model.vector_size  # Отримуємо розмірність GloVe-вектора, наприклад 25
    for sentence in sentences:  # Перебираємо всі речення у наборі даних
        for word, tag in sentence:  # Перебираємо кожне слово та його тег у поточному реченні
            try:  # Починаємо блок, у якому пробуємо отримати GloVe-вектор слова
                # Get the vector for the word  # Отримуємо вектор для поточного слова
                word_vector = glove_model[word]  # Знаходимо у GloVe-моделі числовий вектор слова
            except KeyError:  # Якщо слова немає у словнику GloVe, виникає помилка KeyError
                # Use a zero vector for words not in GloVe vocabulary  # Для невідомого слова створюємо нульовий вектор
                word_vector = np.zeros(vector_size)  # Створюємо вектор із нулів потрібної розмірності
            features.append(word_vector)  # Додаємо отриманий вектор слова до списку ознак
            tags.append(tag)  # Додаємо тег цього слова до списку правильних відповідей
    return np.array(features), tags  # Повертаємо NumPy-масив векторів та список тегів

# Prepare training data using GloVe vectors  # Підготовка навчальних даних за допомогою GloVe-векторів
X_train_glove, y_train_glove = sentences_to_glove_features_and_tags(train_data, glove_vectors)  # Перетворюємо навчальні слова на GloVe-вектори та отримуємо їхні теги

print("Shape of training GloVe features:", X_train_glove.shape)  # Виводимо розмір навчального масиву, наприклад (10000, 25)
print("Number of training tags:", len(y_train_glove))  # Виводимо кількість тегів у навчальному наборі

# Prepare testing data using GloVe vectors  # Підготовка тестових даних за допомогою GloVe-векторів
X_test_glove, y_test_glove = sentences_to_glove_features_and_tags(test_data, glove_vectors)  # Перетворюємо тестові слова на GloVe-вектори та отримуємо їхні теги

print("\nShape of testing GloVe features:", X_test_glove.shape)  # Виводимо розмір тестового масиву
print("Number of testing tags:", len(y_test_glove))  # Виводимо кількість тегів у тестовому наборі

# Train the Logistic Regression model  # Навчаємо модель логістичної регресії
# Increase max_iter for convergence  # Збільшуємо кількість ітерацій, щоб модель мала більше часу для навчання
clf_glove = LogisticRegression(max_iter=1000, verbose=1)  # Створюємо модель Logistic Regression максимум із 1000 ітерацій
clf_glove.fit(X_train_glove, y_train_glove)  # Навчаємо модель на GloVe-векторах та відповідних тегах

# Evaluate the model  # Перевіряємо навчену модель на тестових даних
y_pred_glove = clf_glove.predict(X_test_glove)  # Модель прогнозує тег для кожного слова тестового набору

# Print classification report  # Виводимо детальний звіт про якість класифікації
print("\nClassification Report (GloVe features):")  # Виводимо заголовок звіту

# Ensure target_names are in the same order as the unique tags in y_glove  # Формуємо список усіх унікальних тегів
unique_tags = sorted(list(set(y_train_glove) | set(y_test_glove)))  # Об'єднуємо теги train і test, видаляємо дублікати та сортуємо їх

print(classification_report(y_test_glove, y_pred_glove, zero_division=0, target_names=unique_tags))  # Порівнюємо правильні теги з прогнозами та виводимо precision, recall, f1-score і support
```

## POS tagger за допомогою РНН та word2vec

```python
# I play footbool PAD
# I like going to school

# Вилучення всіх унікальних категорій
all_tags = set()  # Створюємо порожню множину для зберігання всіх унікальних тегів

for sentence in train_data:  # Перебираємо кожне речення з навчального набору даних
    for word, tag in sentence:  # Перебираємо кожне слово та його тег у поточному реченні
        all_tags.add(tag)  # Додаємо тег до множини; повторювані теги автоматично ігноруються

tag_to_ix = {'<PAD>': 0}  # Створюємо словник відповідності "тег → індекс"; PAD отримує індекс 0

for i, tag in enumerate(sorted(list(all_tags))):  # Сортуємо всі унікальні теги та перебираємо їх разом з індексами
    tag_to_ix[tag] = i + 1  # Додаємо тег у словник і призначаємо йому індекс; +1, бо 0 вже зайнятий PAD

ix_to_tag = {ix: tag for tag, ix in tag_to_ix.items()}  # Створюємо зворотний словник для перетворення "індекс → тег"

print("Tag to index mapping:")  # Виводимо заголовок для словника "тег → індекс"
print(tag_to_ix)  # Виводимо відповідність кожного POS-тега його числовому індексу

print("\nIndex to tag mapping:")  # Виводимо заголовок для зворотного словника "індекс → тег"
print(ix_to_tag)  # Виводимо відповідність кожного числового індексу його POS-тегу

print(f"\nNumber of unique tags (including PAD): {len(tag_to_ix)}")  # Виводимо кількість унікальних тегів разом зі спеціальним тегом PAD
```

```python
# Імпортуємо бібліотеку PyTorch для роботи з нейронними мережами та тензорами
import torch

# Імпортуємо базовий клас Dataset і DataLoader для створення та завантаження датасету
from torch.utils.data import Dataset, DataLoader

# Імпортуємо NumPy для роботи з масивами та векторами
import numpy as np

# Імпортуємо WordNetLemmatizer для приведення слів до початкової форми
from nltk.stem import WordNetLemmatizer

# Імпортуємо бібліотеку NLTK для роботи з природною мовою
import nltk


# Завантажуємо словник WordNet, який необхідний для лематизації слів
nltk.download('wordnet')


# Створюємо власний клас датасету для роботи з реченнями та POS-тегами
class NERDataset(Dataset):

    # Конструктор класу, який отримує всі необхідні дані для створення датасету
    def __init__(self, tagged_sentences, glove_model, tag_to_ix, max_sequence_length):

        # Зберігаємо речення з відповідними тегами
        self.tagged_sentences = tagged_sentences

        # Зберігаємо попередньо завантажену GloVe-модель
        self.glove_model = glove_model

        # Зберігаємо словник відповідності тег → числовий індекс
        self.tag_to_ix = tag_to_ix

        # Зберігаємо максимальну довжину речення
        self.max_sequence_length = max_sequence_length

        # Отримуємо розмірність одного GloVe-вектора, наприклад 50, 100 або 300
        self.vector_size = glove_model.vector_size

        # Створюємо лематизатор для приведення слів до початкової форми
        self.lemmatizer = WordNetLemmatizer()


    # Метод __len__ повертає кількість речень у датасеті
    def __len__(self):

        # Повертаємо кількість речень
        return len(self.tagged_sentences)


    # Метод __getitem__ повертає один елемент датасету за його індексом
    def __getitem__(self, idx):

        # Отримуємо речення за переданим індексом
        sentence = self.tagged_sentences[idx]

        # Створюємо порожній список для збереження GloVe-векторів слів
        word_vectors = []

        # Створюємо порожній список для збереження числових індексів тегів
        tag_indices = []


        # Перебираємо всі пари "слово + тег" у поточному реченні
        for word, tag in sentence:

            # Починаємо блок обробки, у якому слово може бути відсутнім у GloVe-словнику
            try:

                # Лематизуємо слово та отримуємо для нього GloVe-вектор
                word_vector = self.glove_model[self.lemmatizer.lemmatize(word)]

            # Якщо лематизованого слова немає у словнику GloVe, виникне помилка KeyError
            except KeyError:

                # Якщо слово відсутнє в GloVe, створюємо нульовий вектор потрібного розміру
                word_vector = np.zeros(self.vector_size)

            # Додаємо отриманий GloVe-вектор до списку векторів
            word_vectors.append(word_vector)

            # Перетворюємо текстовий тег у числовий індекс та додаємо його до списку
            tag_indices.append(
                self.tag_to_ix.get(tag, self.tag_to_ix['<PAD>'])
            )  # Якщо тег відсутній у словнику, використовуємо індекс PAD


        # Створюємо нульову матрицю для GloVe-векторів з максимальною довжиною речення
        padded_vectors = np.zeros(
            (self.max_sequence_length, self.vector_size)
        )

        # Створюємо масив для тегів, заповнений нулями; 0 відповідає тегу PAD
        padded_tags = np.zeros(
            self.max_sequence_length,
            dtype=int
        )


        # Визначаємо фактичну довжину поточного речення
        sequence_length = len(word_vectors)


        # Перевіряємо, чи не перевищує довжина речення максимальну дозволену довжину
        if sequence_length > self.max_sequence_length:

            # Якщо речення занадто довге, обмежуємо його довжину до max_sequence_length
            sequence_length = self.max_sequence_length

            # Обрізаємо GloVe-вектори, залишаючи тільки перші max_sequence_length слів
            word_vectors = word_vectors[:self.max_sequence_length]

            # Обрізаємо POS-теги відповідно до тієї ж довжини
            tag_indices = tag_indices[:self.max_sequence_length]


        # Заповнюємо перші sequence_length рядків матриці GloVe-векторами слів
        padded_vectors[:sequence_length] = np.array(word_vectors)

        # Заповнюємо перші sequence_length позицій масиву відповідними індексами тегів
        padded_tags[:sequence_length] = np.array(tag_indices)


        # Перетворюємо матрицю GloVe-векторів у тензор PyTorch типу float32
        word_vectors_tensor = torch.tensor(
            padded_vectors,
            dtype=torch.float32
        )

        # Перетворюємо індекси тегів у тензор PyTorch типу long
        tag_indices_tensor = torch.tensor(
            padded_tags,
            dtype=torch.long
        )

        # Перетворюємо фактичну довжину речення у тензор PyTorch
        sequence_length_tensor = torch.tensor(
            sequence_length,
            dtype=torch.long
        )

        # Повертаємо GloVe-вектори, теги та реальну довжину речення
        return word_vectors_tensor, tag_indices_tensor, sequence_length_tensor
```

```python
# Імпортуємо основну бібліотеку PyTorch для роботи з тензорами
import torch

# Імпортуємо модулі для створення нейронної мережі
import torch.nn as nn


# Створюємо клас SimpleGRU, який наслідується від базового класу нейронних мереж PyTorch
class SimpleGRU(nn.Module):

    # Конструктор класу, який отримує параметри архітектури GRU
    def __init__(self, input_size, hidden_size, output_size, num_layers=1):

        # Викликаємо конструктор батьківського класу nn.Module
        super(SimpleGRU, self).__init__()

        # Зберігаємо розмір прихованого стану GRU
        self.hidden_size = hidden_size

        # Зберігаємо кількість шарів GRU
        self.num_layers = num_layers

        # Створюємо GRU-шар
        self.gru = nn.GRU(
            input_size,       # Розмір вхідного вектора, наприклад розмір GloVe-вектора
            hidden_size,      # Кількість нейронів у прихованому стані GRU
            num_layers,       # Кількість послідовних шарів GRU
            batch_first=True  # Вказуємо, що першим виміром тензора буде розмір batch
        )

        # Створюємо повнозв'язний шар для перетворення виходу GRU у кількість класів
        self.fc = nn.Linear(
            hidden_size,      # Кількість входів дорівнює розміру прихованого стану GRU
            output_size       # Кількість виходів дорівнює кількості POS-тегів
        )


    # Метод forward описує пряме проходження даних через нейронну мережу
    def forward(self, x):

        # Створюємо початковий прихований стан, заповнений нулями
        h0 = torch.zeros(
            self.num_layers,  # Кількість шарів GRU
            x.size(0),        # Розмір batch, тобто кількість речень
            self.hidden_size  # Розмір прихованого стану
        ).to(x.device)        # Переміщуємо тензор на той самий пристрій, що й вхідні дані


        # Передаємо вхідні дані та початковий прихований стан у GRU
        out, _ = self.gru(x, h0)

        # Після GRU out має форму:
        # (batch_size, seq_length, hidden_size)


        # Передаємо вихід кожного часового кроку через повнозв'язний шар
        out = self.fc(out)

        # Після fc out має форму:
        # (batch_size, seq_length, output_size)


        # Повертаємо результати класифікації для кожного слова в кожному реченні
        return out
```

```python
import torch
import torch.nn as nn
import torch.optim as optim

input_size = glove_vectors.vector_size
hidden_size = 128
output_size = len(tag_to_ix)
num_layers = 1

model = SimpleGRU(input_size, hidden_size, output_size, num_layers)

criterion = nn.CrossEntropyLoss(ignore_index=tag_to_ix['<PAD>']) # Ignore padding in loss calculation
optimizer = optim.Adam(model.parameters(), lr=0.001) # You can adjust the learning rate

print("Model architecture:")
print(model)
```

```python
# Імпортуємо основну бібліотеку PyTorch
import torch

# Імпортуємо модулі для створення нейронних мереж та функцій втрат
import torch.nn as nn

# Імпортуємо модуль optim для роботи з оптимізаторами, наприклад Adam
import torch.optim as optim

# Імпортуємо функції для роботи з послідовностями різної довжини
from torch.nn.utils.rnn import pad_packed_sequence, pack_padded_sequence

# Імпортуємо модуль time для вимірювання часу навчання
import time


# Визначаємо пристрій, на якому буде працювати модель: GPU, якщо CUDA доступна, інакше CPU
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

# Переміщуємо модель на вибраний пристрій — GPU або CPU
model.to(device)


# Встановлюємо кількість епох навчання
num_epochs = 10

# Виводимо інформацію про початок навчання
print(f"Starting training for {num_epochs} epochs...")


# Запускаємо основний цикл навчання; модель пройде через весь train_dataset 10 разів
for epoch in range(num_epochs):

    # Переводимо модель у режим навчання
    model.train()

    # Створюємо змінну для накопичення загальної помилки за всю епоху
    total_loss = 0

    # Запам'ятовуємо час початку поточної епохи
    start_time = time.time()


    # Перебираємо всі batch-и навчального DataLoader
    for i, (batch_vectors, batch_tags, batch_lengths) in enumerate(train_dataloader):

        # Переміщуємо GloVe-вектори, теги та довжини речень на GPU або CPU
        batch_vectors, batch_tags, batch_lengths = (
            batch_vectors.to(device),
            batch_tags.to(device),
            batch_lengths.to(device)
        )


        # Обнуляємо градієнти, які залишилися від попереднього кроку навчання
        optimizer.zero_grad()


        # Передаємо batch GloVe-векторів через нейронну мережу
        outputs = model(batch_vectors)


        # Змінюємо форму виходу моделі з:
        # (batch_size, sequence_length, output_size)
        # на:
        # (batch_size * sequence_length, output_size)
        outputs = outputs.view(-1, outputs.shape[-1])


        # Перетворюємо теги з:
        # (batch_size, sequence_length)
        # у:
        # (batch_size * sequence_length)
        batch_tags = batch_tags.view(-1)


        # Обчислюємо функцію втрат між прогнозами моделі та правильними POS-тегами
        loss = criterion(outputs, batch_tags)


        # Обчислюємо градієнти для всіх параметрів моделі
        loss.backward()


        # Оновлюємо ваги моделі за допомогою оптимізатора
        optimizer.step()


        # Додаємо значення поточної помилки до загальної помилки епохи
        total_loss += loss.item()


        # Кожні 100 batch-ів виводимо поточну інформацію про навчання
        if (i+1) % 100 == 0:

            # Виводимо номер епохи, номер batch-а та поточне значення Loss
            print(
                f'Epoch [{epoch+1}/{num_epochs}], '
                f'Step [{i+1}/{len(train_dataloader)}], '
                f'Loss: {loss.item():.4f}'
            )


    # Запам'ятовуємо час завершення поточної епохи
    end_time = time.time()

    # Обчислюємо середнє значення Loss за всі batch-и поточної епохи
    avg_loss = total_loss / len(train_dataloader)

    # Виводимо інформацію про завершення епохи, її тривалість та середню помилку
    print(
        f"Epoch [{epoch+1}/{num_epochs}] finished in "
        f"{(end_time - start_time):.2f} seconds, "
        f"Average Loss: {avg_loss:.4f}"
    )


# Виводимо повідомлення про завершення всього процесу навчання
print("Training finished.")
```

```python
# Імпортуємо бібліотеку PyTorch для роботи з тензорами та нейронною мережею
import torch

# Імпортуємо classification_report для оцінювання якості класифікації
from sklearn.metrics import classification_report

# Імпортуємо NumPy для роботи з масивами
import numpy as np


# Переводимо модель у режим оцінювання; це вимикає поведінку, яка потрібна тільки під час навчання
model.eval()


# Створюємо порожній список для збереження всіх передбачених моделлю тегів
all_predicted_tags = []

# Створюємо порожній список для збереження всіх правильних тегів
all_true_tags = []


# Вимикаємо обчислення градієнтів, оскільки під час тестування навчання моделі не відбувається
with torch.no_grad():

    # Перебираємо всі batch-и з тестового датасету
    for batch_vectors, batch_tags, batch_lengths in test_dataloader:

        # Переміщуємо GloVe-вектори та правильні теги на вибраний пристрій — CPU або GPU
        batch_vectors, batch_tags = batch_vectors.to(device), batch_tags.to(device)


        # Передаємо тестовий batch через навчену модель
        outputs = model(batch_vectors)


        # Перебираємо кожне речення всередині поточного batch
        for i in range(batch_vectors.size(0)):

            # Отримуємо реальну довжину поточного речення без урахування padding
            sequence_length = batch_lengths[i]


            # Отримуємо передбачення моделі тільки для реальної частини речення
            # argmax вибирає POS-тег із найбільшою оцінкою для кожного слова
            predicted_indices = torch.argmax(
                outputs[i, :sequence_length, :],
                dim=1
            ).cpu().numpy()


            # Отримуємо правильні POS-теги тільки для реальної частини речення
            true_indices = batch_tags[i, :sequence_length].cpu().numpy()


            # Створюємо маску, яка дозволяє виключити позиції з тегом PAD
            non_pad_indices = true_indices != tag_to_ix['<PAD>']


            # Додаємо передбачені теги без PAD до загального списку
            all_predicted_tags.extend(
                predicted_indices[non_pad_indices]
            )


            # Додаємо правильні теги без PAD до загального списку
            all_true_tags.extend(
                true_indices[non_pad_indices]
            )


# Перетворюємо числові індекси передбачених тегів назад у назви POS-тегів
all_predicted_tag_names = [
    ix_to_tag[index] for index in all_predicted_tags
]


# Перетворюємо числові індекси правильних тегів назад у назви POS-тегів
all_true_tag_names = [
    ix_to_tag[index] for index in all_true_tags
]


# Виводимо заголовок перед результатами оцінювання моделі
print("Classification Report for GRU Model:")


# Отримуємо список усіх POS-тегів із навчального набору, виключаючи спеціальний тег PAD
unique_tags_in_train = sorted(
    [tag for tag, ix in tag_to_ix.items() if tag != '<PAD>']
)


# Створюємо та виводимо Classification Report з precision, recall, f1-score та support
print(
    classification_report(
        all_true_tag_names,       # Правильні POS-теги
        all_predicted_tag_names,  # Теги, передбачені GRU
        target_names=unique_tags_in_train  # Назви всіх POS-класів
    )
)
```
